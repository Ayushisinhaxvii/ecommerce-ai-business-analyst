```python
import pandas as pd
import sqlite3
import os
```

```python
db_path = "../../data/processed/ecommerce.db"

conn = sqlite3.connect(db_path)

print("Database connected!")
```

```python
orders = pd.read_csv("../../data/processed/orders_clean.csv")

orders.to_sql(
    "orders",
    conn,
    if_exists="replace",
    index=False
)

print("Orders table created!")
```

```python
customers = pd.read_csv("../../data/processed/customers_clean.csv")
customers.to_sql("customers", conn, if_exists="replace", index=False)

order_items = pd.read_csv("../../data/processed/order_items_clean.csv")
order_items.to_sql("order_items", conn, if_exists="replace", index=False)

payments = pd.read_csv("../../data/processed/payments_clean.csv")
payments.to_sql("payments", conn, if_exists="replace", index=False)

reviews = pd.read_csv("../../data/processed/reviews_clean.csv")
reviews.to_sql("reviews", conn, if_exists="replace", index=False)

products = pd.read_csv("../../data/processed/products_clean.csv")
products.to_sql("products", conn, if_exists="replace", index=False)

sellers = pd.read_csv("../../data/processed/sellers_clean.csv")
sellers.to_sql("sellers", conn, if_exists="replace", index=False)

geolocation = pd.read_csv("../../data/processed/geolocation_clean.csv")
geolocation.to_sql("geolocation", conn, if_exists="replace", index=False)

category_translation = pd.read_csv(
    "../../data/processed/category_translation_clean.csv"
)
category_translation.to_sql(
    "category_translation",
    conn,
    if_exists="replace",
    index=False
)

print("All tables imported!")
```

```python
tables = pd.read_sql(
    "SELECT name FROM sqlite_master WHERE type='table';",
    conn
)

tables
```

```python
for table in tables["name"]:
    count = pd.read_sql(
        f"SELECT COUNT(*) AS row_count FROM {table};",
        conn
    )
    
    print(table, ":", count.loc[0, "row_count"])
```

```python
conn.close()

print("Database connection closed.")
```
